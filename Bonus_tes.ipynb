```python
# ==========================================
# PROGRAM PEMANTAUAN KONSUMSI LISTRIK HARIAN
# ==========================================
# Program ini digunakan untuk menghitung
# konsumsi listrik perangkat elektronik
# dan memperkirakan biaya listrik harian.

print("======================================")
print("   PROGRAM PEMANTAUAN LISTRIK HARIAN")
print("======================================")
```

```python
# Tarif listrik per kWh
tarif_listrik = 1500

# List untuk menyimpan data perangkat
nama_perangkat = []
konsumsi_perangkat = []

total_konsumsi = 0
```

```python
# Membuat variabel dan list
tarif_listrik = 1500
nama_perangkat = []
konsumsi_perangkat = []
total_konsumsi = 0

# Memasukkan jumlah perangkat
try:
    jumlah = int(input("Masukkan jumlah perangkat listrik: "))

    if jumlah <= 0:
        print("Jumlah perangkat harus lebih dari 0!")

    else:
        # Perulangan untuk memasukkan data
        for i in range(jumlah):

            print("\nPerangkat ke-", i + 1)

            nama = input("Nama perangkat: ")

            try:
                daya = float(input("Daya perangkat (Watt): "))
                jam = float(input("Lama penggunaan (jam/hari): "))

                if daya < 0 or jam < 0 or jam > 24:
                    print("Data tidak valid!")
                    continue

                # Menghitung energi listrik
                energi = (daya * jam) / 1000

                # Menyimpan data ke dalam list
                nama_perangkat.append(nama)
                konsumsi_perangkat.append(energi)

                total_konsumsi += energi

            except ValueError:
                print("Input harus berupa angka!")
                continue

        # Menampilkan hasil pemantauan
        print("\n======================================")
        print("          HASIL PEMANTAUAN")
        print("======================================")

        for i in range(len(nama_perangkat)):
            print(nama_perangkat[i], ":",
                  round(konsumsi_perangkat[i], 2), "kWh")

        # Menghitung biaya listrik
        biaya = total_konsumsi * tarif_listrik

        print("--------------------------------------")
        print("Total konsumsi:", round(total_konsumsi, 2), "kWh")
        print("Tarif listrik : Rp", tarif_listrik)
        print("Total biaya   : Rp", round(biaya, 2))

        # Menentukan status penggunaan listrik
        print("--------------------------------------")

        if total_konsumsi < 5:
            print("Status: HEMAT")
            print("Penggunaan listrik masih rendah.")

        elif total_konsumsi <= 10:
            print("Status: NORMAL")
            print("Penggunaan listrik masih terkendali.")

        else:
            print("Status: BOROS")
            print("Penggunaan listrik cukup tinggi!")
            print("Kurangi penggunaan perangkat.")

        print("======================================")
        print("          PROGRAM SELESAI")
        print("======================================")

except ValueError:
    print("Kesalahan! Jumlah perangkat harus berupa angka.")
```
